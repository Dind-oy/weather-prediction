## Weather Predict

```python
# Import libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report, accuracy_score
```

```python
# Load dataset
df = pd.read_csv("seattle-weather.csv")
df.head()
```

```python
df.info()
```

```python
#check for null values
df.isnull().sum()
```

```python
#convert the data type into datetime
df['date'] = pd.to_datetime(df['date'])
```

```python
df.info()
```

```python
df.nunique()
```

```python
df.head()
```

```python
df = df.drop('date', axis=1)
```

```python
df.head()
```

```python
# Separate features (X) and target (y)
X = df[['precipitation', 'temp_max', 'temp_min', 'wind']]
y = df['weather']
```

```python
# Split data into train and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train_scaled = sc.fit_transform(X_train)
X_test_scaled = sc.transform(X_test)
```

### logistic regression

```python
# Train Logistic Regression model
LR = LogisticRegression(max_iter=1000, random_state=0)
LR.fit(X_train_scaled, y_train)

# Make predictions
y_pred_LR = LR.predict(X_test_scaled)

# Evaluate the model
accuracy_LR = accuracy_score(y_test, y_pred_LR)
print(f"accuracy score: {accuracy_LR}")

report_LR = classification_report(y_test, y_pred_LR)
print(f"classification_report:\n {report_LR}")
```

### Random Forest Classifier

```python
# Train Random Forest Classifier
RF = RandomForestClassifier(n_estimators=100, random_state=0)
RF.fit(X_train, y_train)

# Make predictions
y_pred_RF = RF.predict(X_test)

# Evaluate the model
accuracy_RF = accuracy_score(y_test, y_pred_RF)
print(f"accuracy score: {accuracy_RF}")

report_RF = classification_report(y_test, y_pred_RF)
print(f"classification_report:\n {report_RF}")
```

### Decision Tree Classifier

```python
# Train Decision Tree Classifier
DT = DecisionTreeClassifier(random_state=0)
DT.fit(X_train, y_train)

# Make predictions
y_pred_DT = DT.predict(X_test)

# Evaluate the model
accuracy_DT = accuracy_score(y_test, y_pred_DT)
print(f"accuracy score: {accuracy_DT}")

report_DT = classification_report(y_test, y_pred_DT)
print(f"classification_report:\n {report_DT}")
```

### SVM

```python
# Train SVM Classifier
SVM = SVC(kernel='rbf', random_state=0)
SVM.fit(X_train_scaled, y_train)

# Make predictions
y_pred_SVM = SVM.predict(X_test_scaled)

# Evaluate the model
accuracy_SVM = accuracy_score(y_test, y_pred_SVM)
print(f"accuracy score: {accuracy_SVM}")

report_SVM = classification_report(y_test, y_pred_SVM)
print(f"classification_report:\n {report_SVM}")
```

### KNeighbors Classifier

```python
# Train KNN Classifier
KNN = KNeighborsClassifier(n_neighbors=5)  # Default k=5
KNN.fit(X_train_scaled, y_train)

# Make predictions
y_pred_KNN = KNN.predict(X_test_scaled)

# Evaluate the model
accuracy_KNN = accuracy_score(y_test, y_pred_KNN)
print(f"accuracy score: {accuracy_KNN}")

report_KNN = classification_report(y_test, y_pred_KNN)
print(f"classification_report:\n {report_KNN}")
```

### Naive Bayes

```python
# Train Naive Bayes Classifier
NB = GaussianNB()
NB.fit(X_train, y_train)

# Make predictions
y_pred_NB = NB.predict(X_test)

# Evaluate the model
accuracy_NB = accuracy_score(y_test, y_pred_NB)
print(f"accuracy score: {accuracy_NB}")

report_NB = classification_report(y_test, y_pred_NB)
print(f"classification_report:\n {report_NB}")
```

```python
# menampilkan tabel hasil accuracy dan F1 score untuk perbandingan
results = {
    'Models': ['Logistic Regression', 'Random Forest', 'Decision Tree', 'SVM', 'KNeighbors', 'Naive Bayes'],
    'Accuracy': [accuracy_LR, accuracy_RF, accuracy_DT, accuracy_SVM, accuracy_KNN, accuracy_NB]
}

# DataFrame untuk hasil
results_df = pd.DataFrame(results)
results_df
```
